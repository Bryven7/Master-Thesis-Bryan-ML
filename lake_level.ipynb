```python
def create_synthetic_lake_level(df_main, decay_per_eruption=0.15, refill_factor=0.01):
    """
    Simule le niveau du lac basé sur l'historique Pluie vs Eruptions.
    - decay_per_eruption : Combien de mètres on perd par éruption (à calibrer).
    - refill_factor : Combien de mètres on gagne par mm de pluie.
    """
    simulated_level = 1540.0 # Niveau initial (plein)
    levels = []
    
    for index, row in df_main.iterrows():
        # 1. L'impact de la Pluie (Recharge)
        # On suppose que le cratère concentre l'eau
        simulated_level += row['Precipitations_3J'] * refill_factor
        
        # 2. L'impact de l'Éruption (Vidange)
        if row['is_eruption']:
            simulated_level -= decay_per_eruption
            
        # 3. L'Évaporation naturelle (constante)
        simulated_level -= 0.005 
        
        # Bornes physiques (Le lac ne déborde pas à l'infini et ne descend pas sous 0)
        simulated_level = max(1530, min(1542, simulated_level))
        
        levels.append(simulated_level)
        
    return levels
```
